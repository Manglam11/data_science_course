```python
from nltk import bigrams, trigrams, ngrams
```

```python
string = "generative ai & agentic ai is automate the industry & more focus on every domain"
quotes_tokens = nltk.word_tokenize(string)
quotes_tokens
```

```python
len(quotes_tokens)
```

```python
quotes_bigrams = list(nltk.bigrams(quotes_tokens))
quotes_bigrams
```

```python
quotes_trigrams = list(nltk.trigrams(quotes_tokens))
quotes_trigrams
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens, 5))
quotes_ngrams
```

```python
from nltk.stem import PorterStemmer
pst = PorterStemmer()
pst.stem("affection")
```

```python
word_to_stem = ["give", "giving", "given", "gave"]
for word in word_to_stem:
    print(word+ ": "+pst.stem(word))
```

```python
word_to_stem = ["give", "Giving", "given", "gave", "thinking", "loving", "maximum", "manglam"]
for word in word_to_stem:
    print(word+ ": "+pst.stem(word))
```

```python
from nltk.stem import LancasterStemmer
lst = LancasterStemmer()
for word in word_to_stem:
    print(word+ ": "+lst.stem(word))
```

```python
from nltk.stem import SnowballStemmer
sbst = SnowballStemmer("english")
for word in word_to_stem:
    print(word+ ": "+sbst.stem(word))
```

```python
from  nltk.stem import wordnet
from nltk.stem import WordNetLemmatizer
word_lem = WordNetLemmatizer()

for word in word_to_stem:
    print(word+ ": "+word_lem.lemmatize(word))
```
